### Task 4 – Association Rule Learning (Apriori) 

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
df.columns.tolist()
```

```python
basket = (
    df.groupby(["OrderID", "FoodItem"])["FoodItem"]
      .count()
      .unstack()
      .fillna(0)
)

basket.head()
```

```python
basket = (basket > 0).astype(int)

basket.head()
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.02,
    use_colnames=True
)

frequent_itemsets.sort_values(
    by="support",
    ascending=False
).head(20)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.3
)

rules = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
]

rules.sort_values(
    by="lift",
    ascending=False
).head(20)
```

```python
print("Number of frequent itemsets:", len(frequent_itemsets))

frequent_itemsets.sort_values(
    by="support",
    ascending=False
).head(20)
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.005,
    use_colnames=True
)

print("Number of frequent itemsets:", len(frequent_itemsets))

frequent_itemsets.sort_values(
    by="support",
    ascending=False
).head(20)
```

```python
frequent_itemsets[
    frequent_itemsets["itemsets"].apply(len) >= 2
].sort_values(
    by="support",
    ascending=False
).head(20)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.1
)

rules = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
]

rules = rules.sort_values(
    by="lift",
    ascending=False
)

rules.head(20)
```

```python
rules_display = rules.copy()

rules_display["antecedents"] = rules_display["antecedents"].apply(
    lambda x: ", ".join(sorted(x))
)

rules_display["consequents"] = rules_display["consequents"].apply(
    lambda x: ", ".join(sorted(x))
)

rules_display.head(20)
```

### Task 6 – Time Series Forecasting 

```python
print(df["OrderDate"].dtype)
```

```python
monthly_orders = (
    df.set_index("OrderDate")
      .resample("ME")
      .size()
      .reset_index(name="Orders")
)

monthly_orders
```

```python
monthly_revenue = (
    df.set_index("OrderDate")["TotalAmount"]
      .resample("ME")
      .sum()
      .reset_index()
)

monthly_revenue
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_orders,
    x="OrderDate",
    y="Orders",
    marker="o"
)

plt.title("Monthly Order Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_revenue,
    x="OrderDate",
    y="TotalAmount",
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
revenue_ts = monthly_revenue.set_index("OrderDate")["TotalAmount"]

revenue_ts.head()
```

```python
revenue_ts = revenue_ts.asfreq("ME")
```

```python
print("Missing values:", revenue_ts.isna().sum())
```

```python
revenue_ts = revenue_ts.interpolate()
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
model = ARIMA(
    revenue_ts,
    order=(1, 1, 1)
)

model_fit = model.fit()

print(model_fit.summary())
```

```python
forecast = model_fit.forecast(steps=3)

forecast
```

```python
forecast_df = pd.DataFrame({
    "Forecast_Month": forecast.index,
    "Forecast_Revenue": forecast.values
})

forecast_df
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    revenue_ts.index,
    revenue_ts.values,
    marker="o",
    label="Actual Revenue"
)

plt.plot(
    forecast.index,
    forecast.values,
    marker="o",
    linestyle="--",
    label="Forecast Revenue"
)

plt.title("Revenue Forecast Using ARIMA")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

"The dataset contains only 12 monthly observations, while seasonal decomposition with a 12-month period requires at least 24 observations to capture two complete seasonal cycles. Therefore, I avoided forcing seasonal decomposition and focused on the overall monthly trend and ARIMA forecasting."

```python
print("========== NEXT 3 MONTH REVENUE FORECAST ==========")

for date, value in forecast.items():
    print(f"{date.strftime('%Y-%m')}: {value:.2f}")
```

```python
first_revenue = revenue_ts.iloc[0]
last_revenue = revenue_ts.iloc[-1]

percentage_change = (
    (last_revenue - first_revenue) / first_revenue
) * 100

print(f"First Month Revenue: {first_revenue:.2f}")
print(f"Last Month Revenue: {last_revenue:.2f}")
print(f"Overall Change: {percentage_change:.2f}%")
```

```python
print("========== TASK 6 SUMMARY ==========")

print("Historical Trend:")
print(f"- First Month Revenue: ₹{first_revenue:,.2f}")
print(f"- Last Month Revenue: ₹{last_revenue:,.2f}")
print(f"- Overall Change: {percentage_change:.2f}%")

print("\nARIMA Forecast:")
for date, value in forecast.items():
    print(f"- {date.strftime('%B %Y')}: ₹{value:,.2f}")

print("\nObservation:")
print("Historical revenue decreased slightly, while the ARIMA model forecasts relatively stable revenue for the next three months.")
```
